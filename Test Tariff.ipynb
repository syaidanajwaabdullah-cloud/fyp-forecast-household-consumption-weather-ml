```python
"""Sanity tests: the bill calculator reproduces the worked examples published in the Malaysia4U TNB guide
(AFA values as used there). Run: python test_tariff.py"""
import tariff as T
 
CASES = [  # kWh, AFA sen/kWh, exemption limit, expected bill RM
    (200, 0.00, 600, 38.86), (300, 0.00, 600, 65.79), (500, 0.00, 600, 164.74), (600, 0.00, 600, 215.98),
    (900, -0.47, 600, 413.49), (1500, 1.38, 600, 740.14), (2000, 1.38, 600, 1205.02), (3000, 1.38, 600, 1815.60),
]
for kwh, afa, thr, expected in CASES:
    got = T.compute_bill(kwh, afa, thr)["total"]
    assert abs(got - expected) <= 0.02, (kwh, got, expected)
    print(f"{kwh:>5} kWh  RM{got:>9.2f}  (reference RM{expected:>9.2f})  OK")
 
# relief: 700 kWh is exempt from AFA / retail / service tax when the limit is 800
assert T.compute_bill(700, 3.36, 800)["parts"]["AFA"] == 0 and T.compute_bill(700, 3.36, 600)["parts"]["AFA"] > 0
print("All tariff tests passed")
```
