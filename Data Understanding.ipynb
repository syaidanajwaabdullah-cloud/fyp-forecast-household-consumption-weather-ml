# Graphs

```python
# Tetapkan tema visual
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# 1. Muatkan Dataset
df_elec = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\Sem 10 September 2026\Final Year Project 2\dataset\malaysia_daily_2018_to_2026_june30.csv")
df_weather_avg = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\Sem 10 September 2026\Final Year Project 2\dataset\weather_daily_malaysia_average_2018_2026.csv")
df_weather_states = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\Sem 10 September 2026\Final Year Project 2\dataset\weather_daily_all_states_2018_2026.csv")
df_households = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\Sem 10 September 2026\Final Year Project 2\dataset\malaysia_daily_households_2018_2026.csv")

# Tukar kolum date kepada datetime
df_elec['date'] = pd.to_datetime(df_elec['date'])
df_weather_avg['date'] = pd.to_datetime(df_weather_avg['date'])

# --- GRAF 1: Time-Series Line Plot (Penggunaan Elektrik Harian) ---
plt.figure()
sns.lineplot(
    data=df_elec,
    x='date',
    y='consumption_mkwh',
    color='tab:blue',
    linewidth=1.2,
)
plt.title(
    'Malaysia Daily Household Electricity Consumption',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Consumption (MkWh)', fontsize=12)
plt.tight_layout()
plt.show()

# --- GRAF 2: Gabungan Data Cuaca & Elektrik (Scatter Plot dengan Regresi) ---
# Merge dataset elektrik dan cuaca purata harian berdasarkan tarikh
df_merged = pd.merge(df_elec, df_weather_avg, on='date', how='inner')

plt.figure()
sns.regplot(
    data=df_merged,
    x='temperature_C',
    y='consumption_mkwh',
    scatter_kws={'alpha': 0.3, 'color': 'tab:orange'},
    line_kws={'color': 'red', 'linewidth': 2},
)
plt.title(
    'Electricity Consumption vs. Daily Average Temperature',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Temperature (°C)', fontsize=12)
plt.ylabel('Consumption (MkWh)', fontsize=12)
plt.tight_layout()
plt.show()

# --- GRAF 3: Correlation Heatmap Pembolehubah Cuaca ---
plt.figure(figsize=(8, 6))
weather_corr = df_weather_avg[
    ['temperature_C', 'humidity_pct', 'rainfall_mm_per_day', 'wind_speed_m_s']
].corr()
sns.heatmap(weather_corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Weather Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# --- GRAF 4: Perbandingan Suhu Mengikut Negeri (Box Plot) ---
plt.figure(figsize=(14, 6))
sns.boxplot(
    data=df_weather_states,
    x='state',
    y='temperature_C',
    palette='Set2',
    hue='state',
    legend=False,
)
plt.title(
    'Temperature Distribution Across Different States',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('State', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
